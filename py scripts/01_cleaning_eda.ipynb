```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"],errors="coerce")
print(df[df["TotalCharges"].isna()][["customerID","tenure"]])

nan_count=len(df)-df["TotalCharges"].count()
print(nan_count,"is NAN count")

df.loc[df["tenure"]==0,"TotalCharges"]=0
```

```python
for col in ["tenure","MonthlyCharges","TotalCharges"]:
    q1, q3 = df[col].quantile([.25,.75]); iqr = q3 - q1
    n = ((df[col] < q1-1.5*iqr) | (df[col] > q3+1.5*iqr)).sum()
    print(col, n)
```

```python
import pandas as pd

# 1. Convert Churn ("Yes"/"No") to numeric (1/0) so .mean() gives the exact churn rate
df["Churn_num"] = df["Churn"].map({"Yes": 1, "No": 0})

# 2. Create tenure bands (in months)
df["tenure_band"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 60, 72],
    labels=["0-12m", "13-24m", "25-48m", "49-60m", "61-72m"]
)

# 3. List of columns to analyze
cols = [
    "Contract",
    "tenure_band",
    "InternetService",
    "PaymentMethod",
    "SeniorCitizen",
    "TechSupport",
    "OnlineSecurity"
]

# 4. Groupby loop to compute customer count and churn rate (%)
for col in cols:
    summary = (
        df.groupby(col, observed=False)["Churn_num"]
        .agg(Total_Customers="count", Churn_Rate="mean")
        .reset_index()
        
    )
    summary["Churn_Rate"] = (summary["Churn_Rate"] * 100).round(2)
    print(f"=== Churn Rate by {col} ===")
    print(summary.to_string(index=False))
    print("\n")
```
